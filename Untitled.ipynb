#Creating Synthetic data set for the project

```python
import random
import pandas as pd
```

```python
first_names = ["Aarav", "Vihaan", "Aditya", "Ishaan", "Kabir", "Rohan", "Aryan", "Saanvi", "Ananya", "Diya"]
last_names = ["Mehta", "Sharma", "Verma", "Patel", "Iyer", "Nair", "Gupta", "Singh", "Reddy", "Joshi"]
```

```python
def generate_record():

    name = f"{random.choice(first_names)} {random.choice(last_names)}"

    dob = f"{random.randint(1,28):02d}/{random.randint(1,12):02d}/{random.randint(1980,2002)}"

    pan = "".join(random.choices("ABCDEFGHIJKLMNOPQRSTUVWXYZ", k=5)) + \
          str(random.randint(1000,9999)) + \
          random.choice("ABCDEFGHIJKLMNOPQRSTUVWXYZ")

    ocr_text = f"""Name: {name}
DOB: {dob}
PAN: {pan}"""

    transactions = []
    for _ in range(random.randint(2,5)):
        ttype = random.choice(["credit", "debit"])
        mode = random.choice(["salary", "upi", "cash", "card"])
        amount = random.randint(1000, 100000)

        transactions.append({
            "amount": amount,
            "type": ttype,
            "mode": mode
        })

    pep_flag = random.choice([0,0,0,1])        # rare risk
    sanctions_flag = random.choice([0,0,0,0,1])

    return {
        "ocr_text": ocr_text,
        "name": name,
        "matched_name": name,
        "transactions": transactions,
        "pep_flag": pep_flag,
        "sanctions_flag": sanctions_flag
    }
```

```python
dataset = [generate_record() for _ in range(100)]

print(dataset[0])
```
